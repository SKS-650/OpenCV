# Color Detection Using Masks

```python
import numpy as np

# Convert to HSV
hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

# Define color range (Blue)
lower_blue = np.array([100, 150, 0])
upper_blue = np.array([140, 255, 255])

# Create mask
mask = cv2.inRange(hsv, lower_blue, upper_blue)

# Apply mask to original image
result = cv2.bitwise_and(img, img, mask=mask)

cv2.imshow("Original", img)
cv2.imshow("Mask", mask)
cv2.imshow("Result", result)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

#  Track Multiple Colors

```python
# Example: Red color
lower_red = np.array([0, 120, 70])
upper_red = np.array([10, 255, 255])
mask1 = cv2.inRange(hsv, lower_red, upper_red)

lower_red2 = np.array([170, 120, 70])
upper_red2 = np.array([180, 255, 255])
mask2 = cv2.inRange(hsv, lower_red2, upper_red2)

mask_red = mask1 + mask2
red_result = cv2.bitwise_and(img, img, mask=mask_red)

cv2.imshow("Red Color Detection", red_result)
cv2.imshow("Original", img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

# Lesson 9 — Contours (Shape Detection & Object Boundaries)

```python
import cv2

img = cv2.imread("C:/Users/Shyam/Videos/bhoj/test.jpg")
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

cv2.imshow("Original", img)
cv2.imshow("Gray", gray)


# Thresholding
ret, thresh = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)
cv2.imshow("Threshold", thresh)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
contours, hierarchy = cv2.findContours(thresh, cv2.RETR_TREE, cv2.CHAIN_APPROX_SIMPLE)

print(f"Number of contours: {len(contours)}")
```

```python
img_contours = img.copy()
cv2.drawContours(img_contours, contours, -1, (222, 250, 55), 3)  # -1 = draw all

cv2.imshow("Contours", img_contours)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
for cnt in contours:
    x, y, w, h = cv2.boundingRect(cnt)
    cv2.rectangle(img, (x, y), (x+w, y+h), (0, 255, 0), 2)

cv2.imshow("Bounding Boxes", img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
for cnt in contours:
    (x, y), radius = cv2.minEnclosingCircle(cnt)
    center = (int(x), int(y))
    radius = int(radius)
    cv2.circle(img, center, radius, (0, 255, 255), 2)

cv2.imshow("Circles", img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
for cnt in contours:
    area = cv2.contourArea(cnt)
    perimeter = cv2.arcLength(cnt, True)
    print(f"Area: {area}, Perimeter: {perimeter}")
```

# Lesson 10 — Image Histograms

```python
import cv2
from matplotlib import pyplot as plt

img = cv2.imread("C:/Users/Shyam/Videos/bhoj/test.jpg", 0)  # 0 → grayscale

hist = cv2.calcHist([img], [0], None, [256], [0, 256])

plt.plot(hist)
plt.title("Grayscale Histogram")
plt.show()
```

```python
img = cv2.imread("C:/Users/Shyam/Videos/bhoj/test.jpg")

colors = ('b','g','r')
for i, col in enumerate(colors):
    hist = cv2.calcHist([img], [i], None, [256], [0,256])
    plt.plot(hist, color=col)
    plt.xlim([0,256])

plt.title("Color Histogram")
plt.show()
```
