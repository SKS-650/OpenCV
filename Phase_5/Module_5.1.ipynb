# PHASE 5 – VIDEO PROCESSING
### 📘 Module 5.1 – Video Basics

📌 Goal: Process real-time video (file + webcam)


🧠 CORE IDEA (REMEMBER THIS)

A video is just a fast sequence of images (frames)

### 🔹 STEP 1 – READING A VIDEO FILE
📌 ret → True if frame is read successfully
📌 frame → Image (NumPy array)

```python
import cv2
import time

# ================================
# LOAD VIDEO FILE
# ================================

cap = cv2.VideoCapture("video.mp4")

# Check if video opened successfully
if not cap.isOpened():
    print("Error: Cannot open video file")
    exit()

while True:
    ret, frame = cap.read()  # Read one frame

    if not ret:
        print("Video finished")
        break

    cv2.imshow("Video File", frame)

    # Press 'q' to exit
    if cv2.waitKey(25) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```

### 🔹 STEP 2 – WEBCAM ACCESS (VERY IMPORTANT 🔥)

```python
import cv2

# ================================
# OPEN WEBCAM
# ================================

cap = cv2.VideoCapture(0)  # 0 = default camera

if not cap.isOpened():
    print("Error: Cannot access webcam")
    exit()

while True:
    ret, frame = cap.read()

    if not ret:
        break

    cv2.imshow("Webcam Feed", frame)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```

### 🔹 STEP 3 – FRAME-BY-FRAME PROCESSING

📌 Apply OpenCV operations on each frame

Example – Convert Each Frame to Grayscale

```python
import cv2

cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()
    if not ret:
        break

    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    cv2.imshow("Original", frame)
    cv2.imshow("Grayscale Frame", gray)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```

### 🔹 STEP 4 – FPS (FRAMES PER SECOND) CONTROL

- 📌 Control video speed
- 📌 Important for real-time processing


Get FPS from Video

```python
fps = cap.get(cv2.CAP_PROP_FPS)
print("Video FPS:", fps)
```

#### Manual FPS Control (Delay)

```python
import time

cap = cv2.VideoCapture(0)

while True:
    start_time = time.time()

    ret, frame = cap.read()
    if not ret:
        break

    cv2.imshow("FPS Controlled Video", frame)

    # Delay to control FPS (~30 FPS)
    time.sleep(1 / 30)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```

### 🔹 STEP 5 – WRITE VIDEO TO FILE (BONUS 🔥)

```python
import cv2

cap = cv2.VideoCapture(0)

fourcc = cv2.VideoWriter_fourcc(*'XVID')
out = cv2.VideoWriter(
    "output.avi", fourcc, 20.0, (640, 480)
)

while True:
    ret, frame = cap.read()
    if not ret:
        break

    out.write(frame)
    cv2.imshow("Recording", frame)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
out.release()
cv2.destroyAllWindows()
```

# 🧠 IMPORTANT FUNCTIONS (REMEMBER ⭐)
Function ============================  	Purpose
- VideoCapture()	========================  Open video
- read()	==============================  Read frame
- imshow()	============================  Display frame
- waitKey()	============================  Frame delay

# 🧪 PRACTICE (MANDATORY)

- Convert webcam feed to edge detection
- Display FPS on video
- Record grayscale video
- Resize frames before display

### 🧠 COMPLETE PIPELINE
Webcam
 - → Resize frame
 - → Grayscale
 - → Edge detection (Canny)
 - → FPS calculation
 - → Display FPS on frame
 - → Save grayscale video


```python
import cv2
import time

# ================================
# STEP 1: OPEN WEBCAM
# ================================

cap = cv2.VideoCapture(0)

if not cap.isOpened():
    print("Error: Webcam not accessible")
    exit()

# ================================
# STEP 2: VIDEO WRITER (GRAYSCALE)
# ================================
# We still need 3 channels for VideoWriter,
# so grayscale frame will be converted back to BGR

fourcc = cv2.VideoWriter_fourcc(*'XVID')
out = cv2.VideoWriter(
    "grayscale_output.avi",
    fourcc,
    20.0,
    (640, 480)
)

# ================================
# STEP 3: FPS VARIABLES
# ================================

prev_time = 0

# ================================
# STEP 4: MAIN LOOP
# ================================

while True:
    ret, frame = cap.read()
    if not ret:
        break

    # ----------------------------
    # RESIZE FRAME (MANDATORY)
    # ----------------------------
    frame = cv2.resize(frame, (640, 480))

    # ----------------------------
    # CONVERT TO GRAYSCALE
    # ----------------------------
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    # ----------------------------
    # EDGE DETECTION (CANNY)
    # ----------------------------
    edges = cv2.Canny(gray, 40, 50)

    # ----------------------------
    # FPS CALCULATION
    # ----------------------------
    curr_time = time.time()
    fps = 1 / (curr_time - prev_time)
    prev_time = curr_time

    # ----------------------------
    # DISPLAY FPS ON FRAME
    # ----------------------------
    cv2.putText(
        frame,
        f"FPS: {int(fps)}",
        (10, 30),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.8,
        (0, 255, 0),
        2
    )

    # ----------------------------
    # SAVE GRAYSCALE VIDEO
    # Convert gray to BGR before saving
    # ----------------------------
    gray_bgr = cv2.cvtColor(gray, cv2.COLOR_GRAY2BGR)
    out.write(gray_bgr)

    # ----------------------------
    # DISPLAY WINDOWS
    # ----------------------------
    cv2.imshow("Original Webcam (Resized)", frame)
    cv2.imshow("Edge Detection", edges)
    cv2.imshow("Grayscale", gray)

    # Press 'q' to exit
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

# ================================
# CLEANUP
# ================================

cap.release()
out.release()
cv2.destroyAllWindows()
```
