# 📘 Module 1.2 – Color Spaces (EXAMPLES FIRST)

### ✅ Example 1 – Convert Color Image to Grayscale

```python
import cv2

img = cv2.imread("image.jpg")
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

cv2.imshow("Original", img)
cv2.imshow("Grayscale", gray)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

### ✅ Example 2 – Check Channels Difference

```python
print(img.shape)     # (H, W, 3)
print(gray.shape)    # (H, W)
```

### ✅ Example 3 – Split Color Channels (B, G, R)

- 🔵 Blue areas look bright in blue channel
- 🔴 Red areas look bright in red channel

```python
b, g, r = cv2.split(img)

cv2.imshow("Blue Channel", b)
cv2.imshow("Green Channel", g)
cv2.imshow("Red Channel", r)
 
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### ✅ Example 4 – Merge Channels Back

```python
merged = cv2.merge((b, g, r))
cv2.imshow("Merged Image", merged)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

### ✅ Example 5 – Convert BGR → RGB (VERY IMPORTANT)

```python
rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
cv2.imshow("BGR Image", img)
cv2.imshow("RGB Image", rgb)
cv2.imshow("Gray Image", gray)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

### ✅ Example 6 – Convert BGR → HSV (MOST USEFUL)

```python
hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

cv2.imshow("HSV Image", hsv)
cv2.imshow("BGR Image", img)
cv2.imshow("RGB Image", rgb)
cv2.imshow("Gray Image", gray)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### ✅ Example 7 – Extract Only One Color (Real Use Case 🔥)

- 🎯 Detect Blue Color

```python
import numpy as np

hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

lower_white = np.array([100, 100, 100])
upper_white = np.array([250, 250, 255])

mask = cv2.inRange(hsv, lower_white, upper_white)
result = cv2.bitwise_and(img, img, mask=mask)

cv2.imshow("Original Image", img)
cv2.imshow("Mask", mask)
cv2.imshow("White Object", result)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

## ✅ Example 8 – Real-Time Color Detection (Webcam)

```python
import cv2
import numpy as np

cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()
    if not ret:
        break

    hsv = cv2.cvtColor(frame, cv2.COLOR_BGR2HSV)

    lower_red = np.array([0, 120, 70])
    upper_red = np.array([10, 255, 255])

    mask = cv2.inRange(hsv, lower_red, upper_red)
    result = cv2.bitwise_and(frame, frame, mask=mask)

    cv2.imshow("Original", frame)
    cv2.imshow("Red Detection", result)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
