# 📘 Module 1.3 – Drawing Shapes & Text (EXAMPLES FIRST 🔥)

### ✅ Example 1 – Create a Blank Image (Base for Drawing)
- 🧠 This creates a black canvas.

```python
import cv2
import numpy as np

img = np.zeros((500, 500, 3), dtype=np.uint8)
cv2.imshow("Blank Image", img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### ✅ Example 2 – Draw a Line
- cv2.line(image, start_point, end_point, color(BGR), thickness)

```python
img = np.zeros((500, 500, 3), dtype=np.uint8)
cv2.line(img, (50, 450), (450, 450), (25, 2, 200), 3)

cv2.imshow("Line", img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### ✅ Example 3 – Draw a Rectangle

```python
Image = np.zeros((500, 500, 3), dtype=np.uint8)
cv2.rectangle(Image, (50, 100), (450, 250), (255, 0, 0), 2)

cv2.imshow("Rectangle", Image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
img = cv2.imread("image.jpg")
cv2.rectangle(img, (200, 100), (400, 300), (255, 255, 0), 5)

cv2.imshow("Rectangle", img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### ✅ Example 4 – Filled Rectangle

```python
img = cv2.imread("image.jpg")
cv2.rectangle(img, (100, 300), (400, 450), (0, 0, 255), -1)

cv2.imshow("Filled Rectangle", img)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

### ✅ Example 5 – Draw a Circle

```python
img = cv2.imread("image.jpg")
cv2.circle(img, (290, 210), 60, (0, 255, 255), 3)
cv2.circle(img, (290, 210), 200, (220, 255, 0), 3)

cv2.imshow("Circle", img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### ✅ Example 6 – Filled Circle

```python
img = cv2.imread("image.jpg")
cv2.circle(img, (290, 210), 60, (0, 255, 255), -1)  #thickness -1(-ve) for filled circle
cv2.imshow("Filled Circle", img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

## ✅ Example 7 – Put Text on Image (VERY IMPORTANT)

```python
img = cv2.imread("image.jpg")
cv2.putText(
    img,                        
    "OpenCV Rocks!",             # Text to be written
    (50, 350),                   # Starting position of text
    cv2.FONT_HERSHEY_DUPLEX,    # Font family
    1,                           # Font size       
    (0, 255, 255),               # Font color
    2 ,                          # Thickness of the text

)

cv2.imshow("Text", img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### ✅ Example 8 – Draw on Real Image

```python
img = cv2.imread("image.jpg")

cv2.rectangle(img, (190, 110), (400, 300), (0, 255, 0), 3)
cv2.putText(img, "Object", (190, 100), cv2.FONT_HERSHEY_SIMPLEX, 0.8, (0, 255, 0), 2)

cv2.imshow("Annotated Image", img)
cv2.waitKey(0) 
cv2.destroyAllWindows()
```
