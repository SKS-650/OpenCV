# 📘 Module 1.4 – Image Transformations 

### ✅ Example 1 – Read Image & Check Size

```python
import cv2

img = cv2.imread("image.jpg")
print(img.shape)                 # (height, width, channels)
```

### ✅ Example 2 – Resize Image (MOST COMMON)
- Resize to exact size
###### eg.:  cv2.resize(image, (width, height))

```python
img = cv2.imread("image.jpg")
resized = cv2.resize(img, (300, 200))   # Resize to width=300, height=200

cv2.imshow("Original", img)
cv2.imshow("Resized", resized)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

### ✅ Example 3 – Resize with Scale Factor (BETTER)

```python
img = cv2.imread("image.jpg")
resized = cv2.resize(img, None, fx=0.5, fy=0.5)    # Scale down by 50%

cv2.imshow("Original", img)
cv2.imshow("Scaled Down", resized)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### ✅ Example 4 – Flip Image

```python
img = cv2.imread("image.jpg")
horizontal = cv2.flip(img, 1)  # Horizontal flip
vertical = cv2.flip(img, 0)    # Vertical flip
both = cv2.flip(img, -1)       # Both horizontal and vertical flip

cv2.imshow("Original", img)
cv2.imshow("Horizontal Flip", horizontal)
cv2.imshow("Vertical Flip", vertical)
cv2.imshow("Both Flip", both)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

### ✅ Example 5 – Rotate Image (Manual Way)

```python
img = cv2.imread("image.jpg")
(h, w) = img.shape[:2]                             # Get height and width
center = (w//2, h//2)                              # Calculate center of the image

matrix = cv2.getRotationMatrix2D(center, 45, 1.0)  # Rotate 45 degrees
rotated = cv2.warpAffine(img, matrix, (w, h))      # Apply rotation

cv2.imshow("Original", img)
cv2.imshow("Rotated 45 Degree", rotated)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### ✅ Example 6 – Rotate Image (SHORT WAY)

```python
img = cv2.imread("image.jpg")
rotated = cv2.rotate(img, cv2.ROTATE_90_CLOCKWISE)  # Rotate 90 degrees clockwise

cv2.imshow("Original", img)
cv2.imshow("90 Degree Rotation", rotated)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### ✅ Example 7 – Translation (Move Image)

```python
import numpy as np

matrix = np.float32([[1, 0, 100], [0, 1, 50]])   # Shift right by 100 and down by 50
shifted = cv2.warpAffine(img, matrix, (w, h))    # Apply translation

cv2.imshow("Original", img)
cv2.imshow("Shifted Image", shifted)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### ✅ Example 8 – Crop Image (MOST USED)

```python
img = cv2.imread("image.jpg")
crop = img[100:300, 200:400]          # Crop region from (200,100) to (400,300)

cv2.imshow("Original", img)
cv2.imshow("Cropped", crop)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2
img = cv2.imread("image.jpg")
```

```python
# Resize examples
half_width_resized = cv2.resize(img, (img.shape[1]//2, img.shape[0]))  # Resize to half width
half_height_resized = cv2.resize(img, (img.shape[1], img.shape[0]//2))  # Resize to half height

cv2.imshow("Original", img)
cv2.imshow("Half Width Resized", half_width_resized)
cv2.imshow("Half Height Resized", half_height_resized)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Rotation example
rotated_180 =cv2.rotate(img, cv2.ROTATE_180)  # Rotate 180 degrees
cv2.imshow("180 Degree Rotation", rotated_180)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Flipping examples
flipped_horizontal = cv2.flip(img, 1)  # Horizontal flip
flipped_vertical = cv2.flip(img, 0)    # Vertical flip
flipped_both = cv2.flip(img, -1)       # Both horizontal and vertical flip

cv2.imshow("original", img)
cv2.imshow("Flipped Horizontal", flipped_horizontal)
cv2.imshow("Flipped Vertical", flipped_vertical)
cv2.imshow("Flipped Both", flipped_both)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import numpy as np
# Translation example

translation_matrix = np.float32([[1, 0, 50], [0, 1, 30]])  # Translate right by 50 and down by 30 pixels
translated_img = cv2.warpAffine(img, translation_matrix, (img.shape[1], img.shape[0]))

cv2.imshow("Original", img)
cv2.imshow("Translated", translated_img)

cv2.waitKey(0)
cv2.destroyAllWindows()
```
