### ✅ STEP 2: Verify Installation (Example 1)

```python
import cv2
print(cv2.__version__)
```

### ✅ STEP 3: Read & Display Image (Example 2 – MUST RUN)

```python
import cv2

img = cv2.imread("image.jpg")   # put any image in same folder
cv2.imshow("My First Image", img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### ✅ STEP 5: Save Image (Example 3)

```python
cv2.imwrite("saved_image.png", img)
```

### ✅ STEP 6: Image Properties (Example 4 – VERY IMPORTANT)

```python
print(img.shape)    # (height, width, channels)
print(img.dtype)    # usually uint8
```

### ✅ STEP 7: Access Pixel Values (Example 5)

```python
pixel = img[100, 200]
print(pixel)            # [B G R]       
```

### ✅ STEP 8: Modify Pixel (Example 6)

```python
img[100, 200] = [255, 255, 255]  # pure WHITE
cv2.imshow("Pixel Modified", img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### ✅ STEP 10: Region of Interest (ROI) – Crop (Example 8)

```python
roi = img[100:400, 200:500]    #img[row_start : row_end, column_start : column_end]
cv2.imshow("ROI", roi)
cv2.imshow("Pixel Modified", img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
