# 📘 Module 2.2 – Image Transformations

### ✅ Example 1 – Resize Image (Exact Size)

```python
import cv2

img = cv2.imread("image.jpg")

resized = cv2.resize(img, (400, 300))  # width=400, height=300

cv2.imshow("Original", img)
cv2.imshow("Resized", resized)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### ✅ Example 2 – Resize with Scale Factor (Keep Aspect Ratio)

```python
resized = cv2.resize(img, None, fx=0.5, fy=0.5)  # shrink to half

cv2.imshow("Scaled Image", resized)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### ✅ Example 3 – Flip Image

```python
horizontal = cv2.flip(img, 1)  # left-right
vertical = cv2.flip(img, 0)    # up-down
both = cv2.flip(img, -1)       # both axes

cv2.imshow("Original Image", img)
cv2.imshow("Horizontal Flip", horizontal)
cv2.imshow("Vertical Flip", vertical)
cv2.imshow("Both Flip", both)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### ✅ Example 4 – Rotate Image (Using warpAffine)

```python
import numpy as np

(h, w) = img.shape[:2]
center = (w // 2, h // 2)

# Rotate 45 degrees clockwise
matrix = cv2.getRotationMatrix2D(center, 45, 1.0)
rotated = cv2.warpAffine(img, matrix, (w, h))

cv2.imshow("Original Image", img)
cv2.imshow("Rotated Image", rotated)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### ✅ Example 5 – Rotate 90 / 180 / 270 Degrees (Short Way)

```python
rot90 = cv2.rotate(img, cv2.ROTATE_90_CLOCKWISE)
rot180 = cv2.rotate(img, cv2.ROTATE_180)
rot270 = cv2.rotate(img, cv2.ROTATE_90_COUNTERCLOCKWISE)

cv2.imshow("Original Image", img)
cv2.imshow("90 Degree", rot90)
cv2.imshow("180 Degree", rot180)
cv2.imshow("270 Degree", rot270) 

cv2.waitKey(0)
cv2.destroyAllWindows()
```

### ✅ Example 6 – Affine Transformation (Move, Rotate, Skew)

```python
img = cv2.imread("image.jpg")
(h, w) = img.shape[:2]

# Define 3 points in the original image
pts1 = np.float32([[50, 50], [200, 50], [50, 200]])
# Define 3 points in the transformed image
pts2 = np.float32([[10, 100], [200, 50], [100, 250]])

matrix = cv2.getAffineTransform(pts1, pts2)
affine = cv2.warpAffine(img, matrix, (w, h))

cv2.imshow("Affine Transformation", affine)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
