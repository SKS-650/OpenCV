# 📘 Module 2.4 – Bitwise Operations

### ✅ Example 1 – Create Two Images (For Bitwise Ops)

```python
import cv2
import numpy as np

img1 = np.zeros((300, 300, 3), dtype=np.uint8)
img2 = np.zeros((300, 300, 3), dtype=np.uint8)

# Draw shapes
cv2.rectangle(img1, (50, 50), (250, 250), (255, 255, 0), -1)
cv2.circle(img2, (150, 150), 100, (0, 255, 255), -1)

cv2.imshow("Image 1", img1)
cv2.imshow("Image 2", img2)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### ✅ Example 2 – Bitwise AND
- 🔍 Result: Only the overlapping white area is visible.

```python
bitwise_and = cv2.bitwise_and(img1, img2)

cv2.imshow("AND", bitwise_and)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### ✅ Example 3 – Bitwise OR
- 🔍 Result: Union of both shapes

```python
bitwise_or = cv2.bitwise_or(img1, img2)

cv2.imshow("OR", bitwise_or)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### ✅ Example 4 – Bitwise XOR
- 🔍 Result: Only non-overlapping regions remain.

```python
bitwise_xor = cv2.bitwise_xor(img1, img2)

cv2.imshow("XOR", bitwise_xor)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### ✅ Example 5 – Bitwise NOT
- 🔍 Result: Colors are inverted.

```python
bitwise_not = cv2.bitwise_not(img1)

cv2.imshow("NOT", bitwise_not)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

## ✅ Example 6 – Masking (MOST IMPORTANT 🔥)

#### 🎯 Extract object from image
🔥 Used in:

- background removal
- object extraction
- segmentation

```python
img = cv2.imread("image.jpg")
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

_, mask = cv2.threshold(gray, 200, 255, cv2.THRESH_BINARY)

result = cv2.bitwise_and(img, img, mask=mask)

cv2.imshow("Original", img)
cv2.imshow("Mask", mask)
cv2.imshow("Masked Result", result)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

### ✅ Example 7 – Color Masking using HSV (Real Use 🔥)

```python
hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

lower_blue = np.array([100, 150, 50])
upper_blue = np.array([140, 255, 255])

mask = cv2.inRange(hsv, lower_blue, upper_blue)
blue_object = cv2.bitwise_and(img, img, mask=mask)

cv2.imshow("Original Image", img) 
cv2.imshow("Blue Mask", mask)
cv2.imshow("Blue Object", blue_object)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### ✅ Example 8 – Image Blending (Overlay Images)

```python
img1 = cv2.imread("image1.jpg")
img2 = cv2.imread("image2.jpg")

img2 = cv2.resize(img2, (img1.shape[1], img1.shape[0]))

blended = cv2.addWeighted(img1, 0.8, img2, 0.2, 0) # alpha=0.5, beta=0.5, gamma=0

cv2.imshow("Image 1", img1)
cv2.imshow("Image 2", img2)
cv2.imshow("Blended Image", blended)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

# 📌 PHASE 2 – Mini Project
### 🛠️ Create a Basic Image Editor
🎯 PROJECT GOAL
Build a basic image editor that can:

- Load image
- Resize
- Rotate
- Flip
- Convert color spaces
- Draw shapes & text
- Blend two images
- Save output

```python
# Basic image editor using OpenCV
import cv2
import numpy as np

# -----------------------------
# LOAD IMAGE
# -----------------------------
img = cv2.imread("image1.jpg")

if img is None:
    print("Error: image1.jpg not found")
    exit()

cv2.imshow("Original Image", img)
cv2.waitKey(0)

# -----------------------------
# RESIZE
# -----------------------------
resized = cv2.resize(img, (400, 300))
cv2.imshow("Resized Image", resized)
cv2.waitKey(0)

# -----------------------------
# ROTATE
# -----------------------------
(h, w) = img.shape[:2]
center = (w // 2, h // 2)
rot_matrix = cv2.getRotationMatrix2D(center, 45, 1.0)
rotated = cv2.warpAffine(img, rot_matrix, (w, h))

cv2.imshow("Rotated Image", rotated)
cv2.waitKey(0)

# -----------------------------
# FLIP
# -----------------------------
flipped = cv2.flip(img, 1)
cv2.imshow("Flipped Image", flipped)
cv2.waitKey(0)

# -----------------------------
# COLOR CONVERSIONS
# -----------------------------
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

cv2.imshow("Grayscale", gray)
cv2.imshow("HSV", hsv)
cv2.waitKey(0)

# -----------------------------
# DRAW SHAPES & TEXT
# -----------------------------
draw = img.copy()

cv2.rectangle(draw, (50, 50), (250, 200), (0, 255, 0), 2)
cv2.circle(draw, (350, 250), 40, (255, 0, 0), -1)
cv2.putText(
    draw,
    "Basic Image Editor",
    (50, 450),
    cv2.FONT_HERSHEY_SIMPLEX,
    1,
    (0, 0, 255),
    2
)

cv2.imshow("Drawing", draw)
cv2.waitKey(0)

# -----------------------------
# MASKING (Threshold Based)
# -----------------------------
_, mask = cv2.threshold(gray, 200, 255, cv2.THRESH_BINARY)
masked = cv2.bitwise_and(img, img, mask=mask)

cv2.imshow("Mask", mask)
cv2.imshow("Masked Image", masked)
cv2.waitKey(0)

# -----------------------------
# BITWISE OPERATIONS DEMO
# -----------------------------
canvas1 = np.zeros((300, 300, 3), dtype=np.uint8)
canvas2 = np.zeros((300, 300, 3), dtype=np.uint8)

cv2.rectangle(canvas1, (50, 50), (250, 250), (255, 255, 255), -1)
cv2.circle(canvas2, (150, 150), 100, (255, 255, 255), -1)

bit_and = cv2.bitwise_and(canvas1, canvas2)
bit_or = cv2.bitwise_or(canvas1, canvas2)
bit_xor = cv2.bitwise_xor(canvas1, canvas2)
bit_not = cv2.bitwise_not(canvas1)

cv2.imshow("AND", bit_and)
cv2.imshow("OR", bit_or)
cv2.imshow("XOR", bit_xor)
cv2.imshow("NOT", bit_not)
cv2.waitKey(0)

# -----------------------------
# IMAGE BLENDING (OPTIONAL)
# -----------------------------
img2 = cv2.imread("image2.jpg")

if img2 is not None:
    img2 = cv2.resize(img2, (img.shape[1], img.shape[0]))
    blended = cv2.addWeighted(img, 0.6, img2, 0.4, 0)
    cv2.imshow("Blended Image", blended)
    cv2.imwrite("final_output.jpg", blended)
else:
    blended = draw
    cv2.imwrite("final_output.jpg", blended)

cv2.waitKey(0)
cv2.destroyAllWindows()

print("Project completed. Output saved as final_output.jpg")
```
