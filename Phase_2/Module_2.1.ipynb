# 🟡 PHASE 2 – BASIC IMAGE PROCESSING

### 📘 Module 2.1 – Color Spaces

##### ✅ Example 1 – BGR vs RGB (MOST COMMON CONFUSION)

```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread("image.jpg")

# OpenCV uses BGR
cv2.imshow("BGR Image", img)
cv2.waitKey(0)
cv2.destroyAllWindows()

# Convert to RGB for correct display in matplotlib
rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.imshow(rgb)
plt.title("RGB Image")
plt.axis("off")
plt.show()
```

#### ✅ Example 2 – Convert Image to Grayscale

```python
img = cv2.imread("image.jpg")
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

cv2.imshow("Grayscale Image", gray)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

#### ✅ Example 3 – Why Grayscale? (Edge Example)
- 📌 Grayscale reduces computation and noise.

```python
edges1 = cv2.Canny(gray, 100, 200)
edges2 = cv2.Canny(gray, 50, 150)
edges3 = cv2.Canny(gray, 30, 100)
edges4 = cv2.Canny(gray, 150, 300)

cv2.imshow("Edges with thresholds 100 and 200", edges1)
cv2.imshow("Edges with thresholds 50 and 150", edges2)
cv2.imshow("Edges with thresholds 30 and 100", edges3)
cv2.imshow("Edges with thresholds 150 and 300", edges4)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

#### ✅ Example 4 – Convert BGR → HSV (VERY IMPORTANT 🔥)

```python
hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

cv2.imshow("BGR Image", img)
cv2.imshow("HSV Image", hsv)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

#### ✅ Example 5 – Detect Color Using HSV (REAL USE 🔥)

- 🎯 Detect Green Object

```python
import numpy as np

hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

lower_green = np.array([35, 50, 50])
upper_green = np.array([85, 255, 255])

mask = cv2.inRange(hsv, lower_green, upper_green)
result = cv2.bitwise_and(img, img, mask=mask)

cv2.imshow("Original Image", img)
cv2.imshow("Mask", mask)
cv2.imshow("Green Detection", result)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

#### ✅ Example 6 – LAB Color Space (Lighting Robust 🔥)
- 📌 LAB separates lightness from color
→ very useful in poor lighting conditions

```python
lab = cv2.cvtColor(img, cv2.COLOR_BGR2LAB)  

cv2.imshow("LAB Image", lab)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

#### ✅ Example 7 – Compare All Color Spaces Side-by-Side 

```python
cv2.imshow("BGR", img)
cv2.imshow("Gray", gray)
cv2.imshow("HSV", hsv)
cv2.imshow("LAB", lab)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

## 🧪 PRACTICE (IMPORTANT)

Try this yourself:

- Detect white color using HSV
- Convert LAB → BGR
- Use grayscale image and apply Canny
- Show only Hue channel of HSV

```python
# detect white color using HSV color space
import numpy as np

img = cv2.imread("image.jpg")
hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

lower_white = np.array([0, 0, 200])
upper_white = np.array([180, 25, 255])

mask = cv2.inRange(hsv, lower_white, upper_white)
result = cv2.bitwise_and(img, img, mask=mask)

cv2.imshow("Original Image", img)
cv2.imshow("Mask for White Color", mask)
cv2.imshow("White Color Detection", result)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# convert BGR to LAB and back to BGR
img = cv2.imread("image.jpg")
lab = cv2.cvtColor(img, cv2.COLOR_BGR2LAB)
bgr = cv2.cvtColor(lab, cv2.COLOR_LAB2BGR)

cv2.imshow("Original BGR Image", img)
cv2.imshow("Converted BGR from LAB", bgr)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# use gray scale for Canny edge detection
img = cv2.imread("image.jpg")
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
edges1 = cv2.Canny(gray, 100, 200)
cv2.imshow("Edges with thresholds 100 and 200", edges1)
cv2.waitKey(0)
cv2.destroyAllWindows()

```

```python
# showing hue channel from HSV image

img = cv2.imread("image.jpg")
hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
hue_channel = hsv[:, :, 0]
cv2.imshow("Hue Channel", hue_channel)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
