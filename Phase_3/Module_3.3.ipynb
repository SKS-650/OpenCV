# 📘 Module 3.3 – Edge Detection

- Sobel
- Laplacian
- Canny edge detection

📌 Goal: Understand how OpenCV detects boundaries & shapes

🧠 One-Line Concept

- Edges = sudden intensity change
- OpenCV finds edges using derivatives (gradients)

### 🔹 STEP 0 – Base Setup (MANDATORY)

```python
import cv2
import numpy as np

Image = cv2.imread("image.jpg")
img = cv2.resize(Image, (600, 400))    # 600 is width, 400 is height
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

cv2.imshow("Original", img)
cv2.imshow("Gray", gray)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### 🔹 WHY EDGE DETECTION MATTERS

Used in:

- Document scanning
- Object detection
- Shape recognition
- OCR
- Face recognition

### ✅ 1️⃣ SOBEL EDGE DETECTION

- 📌 Detects edges in X or Y direction
- 📌 First-order derivative

##### Example 1 – Sobel X (Vertical Edges)

```python
sobel_x = cv2.Sobel(gray, cv2.CV_64F, 1, 0, ksize=3) # Sobel operator in x direction and kernel size 3
sobel_x = cv2.convertScaleAbs(sobel_x) # Convert back to uint8

cv2.imshow("Sobel X", sobel_x)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

#### Example 2 – Sobel Y (Horizontal Edges)

```python
sobel_y = cv2.Sobel(gray, cv2.CV_64F, 0, 1, ksize=3)
sobel_y = cv2.convertScaleAbs(sobel_y)

cv2.imshow("Sobel Y", sobel_y)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

#### Example 3 – Sobel Combined

```python
sobel_combined = cv2.addWeighted(sobel_x, 0.5, sobel_y, 0.5, 0) # Combine both gradients

cv2.imshow("Sobel Combined", sobel_combined)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

--- 
### ✅ 2️⃣ LAPLACIAN EDGE DETECTION

- 📌 Detects edges in all directions
- 📌 Second-order derivative
- 📌 Sensitive to noise

##### Example 4 – Laplacian

```python
laplacian = cv2.Laplacian(gray, cv2.CV_64F)
laplacian = cv2.convertScaleAbs(laplacian)

cv2.imshow("Laplacian", laplacian)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### ⚠️ Laplacian + Blur (BEST PRACTICE)

```python
blur = cv2.GaussianBlur(gray, (5,5), 0)
laplacian_blur = cv2.Laplacian(blur, cv2.CV_64F)
laplacian_blur = cv2.convertScaleAbs(laplacian_blur)

cv2.imshow("Laplacian After Blur", laplacian_blur)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

---
### ✅ 3️⃣ CANNY EDGE DETECTION (🔥 MOST IMPORTANT 🔥)

📌 Uses:

- Gaussian blur
- Gradient calculation
- Non-maximum suppression
- Hysteresis thresholding

📌 Industry standard

##### Example 5 – Canny Basic

```python
edges = cv2.Canny(gray, 100, 200)     # Canny Edge Detection with thresholds 100 and 200

cv2.imshow("Canny Edges", edges)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

#### Example 6 – Canny with Blur (BEST)

```python
blur = cv2.GaussianBlur(gray, (5,5), 0)
edges = cv2.Canny(blur, 50, 150)

cv2.imshow("Blurred", blur)
cv2.imshow("Canny After Blur", edges)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### 🔍 EDGE COMPARISON (VERY IMPORTANT)

```python
cv2.imshow("Sobel", sobel_combined)
cv2.imshow("Laplacian", laplacian)
cv2.imshow("Canny", edges)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

## 🔥 MINI PIPELINE (DOCUMENT SCAN BASE)

```python
blur = cv2.GaussianBlur(gray, (5,5), 0)
edges = cv2.Canny(blur, 75, 200)

cv2.imshow("Edges for Document", edges)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
