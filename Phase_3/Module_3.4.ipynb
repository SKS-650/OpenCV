# 📘 Module 3.4 – Morphological Operations

- ✔ Erosion
- ✔ Dilation
- ✔ Opening
- ✔ Closing
- ✔ Morphological Gradient

📌 Goal: Shape-based image processing (very important for document scanner)

##### 🧠 One-Line Concept

Morphology = modify shapes using a structuring element

#### 🔹 STEP 0 – Base Setup (MANDATORY)

```python
import cv2
import numpy as np

Image = cv2.imread("image.jpg")
img = cv2.resize(Image, (600, 400))    # 600 is width, 400 is height
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Binary image needed for morphology
_, binary = cv2.threshold(gray, 150, 255, cv2.THRESH_BINARY)

cv2.imshow("Original", img)
cv2.imshow("Binary", binary)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### 🔹 STRUCTURING ELEMENT (KERNEL)
- 📌 Kernel decides shape & strength



```python
kernel = np.ones((5, 5), np.uint8)
```

### ✅ 1️⃣ EROSION

📌 Shrinks white regions
📌 Removes small noise

#### Example 1 – Erosion

```python
erosion = cv2.erode(binary, kernel, iterations=1)

cv2.imshow("Erosion", erosion)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### ✅ 2️⃣ DILATION

- 📌 Expands white regions
- 📌 Fills gaps

#### Example 2 – Dilation

```python
dilation = cv2.dilate(binary, kernel, iterations=1)

cv2.imshow("Dilation", dilation)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### ✅ 3️⃣ OPENING (EROSION ➝ DILATION)

- 📌 Removes noise
- 📌 Keeps object shape

##### Example 3 – Opening

```python
opening = cv2.morphologyEx(binary, cv2.MORPH_OPEN, kernel)

cv2.imshow("Opening", opening)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### ✅ 4️⃣ CLOSING (DILATION ➝ EROSION)

- 📌 Fills holes
- 📌 Connects broken parts

#### Example 4 – Closing

```python
closing = cv2.morphologyEx(binary, cv2.MORPH_CLOSE, kernel)

cv2.imshow("Closing", closing)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### ✅ 5️⃣ MORPHOLOGICAL GRADIENT

📌 Edge = dilation − erosion

#### Example 5 – Morphological Gradient

```python
gradient = cv2.morphologyEx(binary, cv2.MORPH_GRADIENT, kernel)

cv2.imshow("Morphological Gradient", gradient)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

## 🔍 MORPHOLOGY COMPARISON (IMPORTANT)

```python

cv2.imshow("Original Binary", binary)
cv2.imshow("Erosion", erosion)
cv2.imshow("Dilation", dilation)
cv2.imshow("Opening", opening)
cv2.imshow("Closing", closing)
cv2.imshow("Gradient", gradient)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

## 🔥 MINI PIPELINE – DOCUMENT SCANNER BASE

```python
blur = cv2.GaussianBlur(gray, (5,5), 0)
_, thresh = cv2.threshold(blur, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)

kernel = np.ones((5,5), np.uint8)
closed = cv2.morphologyEx(thresh, cv2.MORPH_CLOSE, kernel)

edges = cv2.Canny(closed, 75, 200)

cv2.imshow("Final Edges", edges)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

# 🟡 PHASE 3 – MINI PROJECT
- 📌 Document Scanner + Noise Removal Pipeline

🎯 WHAT THIS MINI PROJECT DOES

- Removes noise
- Detects document edges
- Cleans the document
- Prepares it for OCR / scanning

📌 This is the core logic used in real scanning apps

##### 🧠 PIPELINE OVERVIEW (VERY IMPORTANT)
Image
 - Grayscale
 - Blur (Noise Removal)
 - Threshold (Binary)
 - Morphology (Clean edges)
 - Canny Edge Detection
 - Final scanned edges

```python
import cv2
import numpy as np

# ================================
# STEP 1: LOAD IMAGE
# ================================

Image = cv2.imread("image.jpg")
img = cv2.resize(Image, (600, 400))    # 600 is width, 400 is height
if img is None:
    print("Image not found!")
    exit()

cv2.imshow("Original Image", img)
cv2.waitKey(0)

# ================================
# STEP 2: CONVERT TO GRAYSCALE
# ================================

gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
cv2.imshow("Grayscale", gray)
cv2.waitKey(0)

# ================================
# STEP 3: NOISE REMOVAL (BLUR)
# Gaussian Blur removes sensor noise
# ================================

blur = cv2.GaussianBlur(gray, (5, 5), 0)
cv2.imshow("Blurred Image", blur)
cv2.waitKey(0)

# ================================
# STEP 4: THRESHOLDING (OTSU)
# Converts image to black & white
# ================================

_, thresh = cv2.threshold(
    blur, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU
)
cv2.imshow("Binary Image", thresh)
cv2.waitKey(0)

# ================================
# STEP 5: MORPHOLOGICAL OPERATIONS
# Closing fills gaps in document edges
# ================================

kernel = np.ones((5, 5), np.uint8)
closed = cv2.morphologyEx(thresh, cv2.MORPH_CLOSE, kernel)

cv2.imshow("After Morphology (Closing)", closed)
cv2.waitKey(0)

# ================================
# STEP 6: EDGE DETECTION (CANNY)
# Extracts clean document boundaries
# ================================

edges = cv2.Canny(closed, 75, 200)

cv2.imshow("Final Document Edges", edges)
cv2.waitKey(0)

# ================================
# CLEANUP
# ================================

cv2.destroyAllWindows()
```
