# 🟡 PHASE 3 – INTERMEDIATE IMAGE PROCESSING

## 📘 Module 3.1 – Image Thresholding

- Simple threshold
- Adaptive threshold
- Otsu thresholding

### 🔹 STEP 0 – Load Image & Convert to Grayscale (MANDATORY)

```python
import cv2

image = cv2.imread("image.jpg")
img = cv2.resize(image, (600, 400))    # 600 is width, 400 is height
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

cv2.imshow("Original", img)
cv2.imshow("Grayscale", gray)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### Example 1 – Binary Threshold
🔍 What happens:

- Pixel > 127 → White (255)
- Pixel ≤ 127 → Black (0)

```python
_, thresh = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)   #threshold(source image, threshold value, max value, thresholding type)

cv2.imshow("Binary Threshold", thresh)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### Example 2 – Binary Inverted
- just opposite


```python
_, thresh_inv = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY_INV) # Inverted Binary Threshold

cv2.imshow("Binary Inverted", thresh_inv)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### Example 3 – Truncate Threshold

```python
_, thresh_trunc = cv2.threshold(gray, 127, 255, cv2.THRESH_TRUNC) # pixels above the threshold are set to the threshold value, below are unchanged

cv2.imshow("Truncate Threshold", thresh_trunc)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### Example 4 – To Zero Threshold

```python
_, thresh_zero = cv2.threshold(gray, 127, 255, cv2.THRESH_TOZERO) # pixels above the threshold are kept unchanged, below are set to zero

cv2.imshow("ToZero Threshold", thresh_zero)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### Example 5 – Adaptive Mean Threshold

```python
adaptive_mean = cv2.adaptiveThreshold(
    gray,
    255,
    cv2.ADAPTIVE_THRESH_MEAN_C,
    cv2.THRESH_BINARY,
    11,
    2
)
# adaptiveThreshold(source image, max value, adaptive method, threshold type, block size, constant subtracted from mean)

cv2.imshow("Adaptive Mean Threshold", adaptive_mean)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### Example 6 – Adaptive Gaussian Threshold
🔥 Used in:

- document scanning
- number plate detection
- handwriting recognition

```python
adaptive_gaussian = cv2.adaptiveThreshold(
    gray,
    255,
    cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
    cv2.THRESH_BINARY,
    11,
    2
)

cv2.imshow("Adaptive Gaussian Threshold", adaptive_gaussian)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### Example 7 – Otsu Threshold
- 📌 No threshold value given by us → OpenCV decides.
- OpenCV finds best threshold automatically.

```python
_, otsu = cv2.threshold(
    gray,
    0,
    255,
    cv2.THRESH_BINARY + cv2.THRESH_OTSU
)

cv2.imshow("Otsu Threshold", otsu)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

## Compare All Thresholds (IMPORTANT)

```python
cv2.imshow("Gray", gray)
cv2.imshow("Binary", thresh)
cv2.imshow("Adaptive", adaptive_gaussian)
cv2.imshow("Otsu", otsu)

cv2.waitKey(0)
cv2.destroyAllWindows()
```
