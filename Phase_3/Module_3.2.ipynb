# 📘 Module 3.2 – Smoothing & Blurring

- Gaussian blur
- Median blur
- Bilateral filter
- Noise types



- 📌 Goal: Remove noise & prepare image for edge detection and scanning

🧠 One-Line Concept

- Blurring = Noise reduction without destroying important structures

### 🔹 STEP 0 – Import & Load Image (BASE SETUP)

```python
import cv2

Image = cv2.imread("image.jpg")
img = cv2.resize(Image, (600, 400))    # 600 is width, 400 is height
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

cv2.imshow("Original", img)
cv2.imshow("Gray", gray)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

##### 🔹 NOISE TYPES (IMPORTANT FOR EXAMS + PROJECT)
Noise Type  :                    	Cause
- Gaussian Noise	 :               Sensor, low light
- Salt & Pepper	     :             Transmission error
- Speckle	          :              Medical / radar
- Poisson	            :            Photon noise

### ✅ 1️⃣ GAUSSIAN BLUR

- 📌 Removes Gaussian noise
- 📌 Uses weighted average

#### Example 1 – Gaussian Blur

```python
gaussian = cv2.GaussianBlur(gray, (5, 5), 0) # (source image, kernel size, sigma)  

cv2.imshow("Gaussian Blur", gaussian)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

##### Try More Blur

```python
gaussian_strong = cv2.GaussianBlur(gray, (15, 15), 0)

cv2.imshow("Gaussian Strong Blur", gaussian_strong)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### ✅ 2️⃣ MEDIAN BLUR (VERY IMPORTANT 🔥)

- 📌 Best for Salt & Pepper noise
- 📌 Preserves edges better

##### Example 2 – Median Blur
- 📌 Kernel size must be odd

```python
median = cv2.medianBlur(gray, 5)

cv2.imshow("Median Blur", median)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### ✅ 3️⃣ BILATERAL FILTER (ADVANCED 🔥🔥)

- 📌 Removes noise
- 📌 Keeps edges sharp
- 📌 Used in document scanning

##### Example 3 – Bilateral Filter

```python
bilateral = cv2.bilateralFilter(gray, 9, 75, 75) #9 is diameter of pixel neighborhood, 75 is sigmaColor, 75 is sigmaSpace

cv2.imshow("Bilateral Filter", bilateral)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### 🔍 COMPARISON (VERY IMPORTANT)

```python
cv2.imshow("Original Gray", gray)
cv2.imshow("Gaussian", gaussian)
cv2.imshow("Median", median)
cv2.imshow("Bilateral", bilateral)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

# 🔥 MINI PIPELINE (BLUR → THRESHOLD)

```python
blur = cv2.GaussianBlur(gray, (5,5), 0)
_, binary = cv2.threshold(blur, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)

cv2.imshow("Blurred", blur)
cv2.imshow("Binary After Blur", binary)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
