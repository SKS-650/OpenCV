# 🟡 PHASE 4 – CONTOURS & SHAPE ANALYSIS
## 📘 Module 4.1 – Contours

📌 Goal: Extract shapes & objects from images

🧠 ONE-LINE IDEA (REMEMBER THIS)

- Contours are continuous boundaries of objects having the same color/intensity

📌 OpenCV finds contours from binary images, not color images.

### STEP 0 – BASE SETUP (MANDATORY)

```python
import cv2
import numpy as np

# Load image
Image = cv2.imread("image.jpg")
img = cv2.resize(Image, (450, 300))
if img is None:
    print("Image not found")
    exit()

cv2.imshow("Original", img)
cv2.waitKey(0)

# Convert to grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Convert to binary
_, binary = cv2.threshold(gray, 150, 255, cv2.THRESH_BINARY)

cv2.imshow("Binary Image", binary)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

##### ✅ WHAT ARE CONTOURS? (VERY SHORT THEORY)

- ✔ Curve joining boundary points
- ✔ Detected from white objects on black background

######  Used for:

- Shape detection
- Object counting
- Document scanning

### ✅ 1️⃣ CONTOUR DETECTION

📌 MOST IMPORTANT FUNCTION

```python
contours, hierarchy = cv2.findContours(
    binary,
    cv2.RETR_EXTERNAL,
    cv2.CHAIN_APPROX_SIMPLE
)

print("Total contours found:", len(contours))
```

### ✅ 2️⃣ DRAWING CONTOURS

```python
# Copy image to draw on
contour_img = img.copy()

cv2.drawContours(
    contour_img,    # Image to draw on
    contours,       # Contours to draw
    -1,             # Draw all contours
    (0, 255, 0),    # Green color
    2               # Thickness
)

cv2.imshow("Original", img)
cv2.imshow("Contours Drawn", contour_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### 🔍 DRAW SINGLE CONTOUR (IMPORTANT)

```python
single = img.copy()

cv2.drawContours(single, contours, 0, (255, 0, 0), 1)

cv2.imshow("First Contour", single)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
